```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
```

```python
text = """
Information retrieval is an important field of computer science.
It helps users find useful information from large collections of documents.
The system processes the documents and retrieves relevant information.
"""

print("Original Text:")
print(text)
```

```python
tokens = word_tokenize(text)

print("Tokens:")
print(tokens)
```

```python
stop_words = set(stopwords.words('english'))

filtered_words = []

for word in tokens:
    if word.lower() not in stop_words:
        filtered_words.append(word)

print("After Stop-Word Removal:")
print(filtered_words)
```

```python
ps = PorterStemmer()

stemmed_words = []

for word in filtered_words:
    if word.isalpha():
        stemmed_words.append(ps.stem(word))
    else:
        stemmed_words.append(word)

print("After Stemming:")
print(stemmed_words)
print("Aditya Rade   BEAD23167")
```
