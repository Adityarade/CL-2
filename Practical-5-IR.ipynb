Step 1: Import NetworkX

```python
import networkx as nx
```

Step 2: Create the Graph

```python
G = nx.barabasi_albert_graph(60, 41)

print("Number of Nodes:", G.number_of_nodes())
print("Number of Edges:", G.number_of_edges())
```

Step 3: Calculate PageRank

```python
pr = nx.pagerank(G, alpha=0.4)

print("PageRank Values:")
print(pr)
```

Step 4: Display PageRank in Descending Order

```python
sorted_pr = sorted(pr.items(), key=lambda x: x[1], reverse=True)

print("PageRank in Descending Order:\n")

for node, score in sorted_pr:
    print(f"Node {node}: {score:.6f}")
```

Step 5: Find the Highest PageRank Node

```python
highest_node = max(pr, key=pr.get)

print("\nHighest PageRank Node:", highest_node)
print("Highest PageRank Score:", pr[highest_node])
```

Step 6: Visualize the Graph

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
nx.draw(G, with_labels=True, node_size=500)
plt.title("Barabasi-Albert Graph")
plt.show()
print("Aditya Rade   BEAD23167")

```
