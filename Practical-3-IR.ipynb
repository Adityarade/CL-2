```python
import numpy as np
import pandas as pd

from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.estimators import MaximumLikelihoodEstimator

# Load dataset
heartDisease = pd.read_csv("heart.csv")

# Clean column names
heartDisease.columns = heartDisease.columns.str.strip()

# Handle missing values
heartDisease = heartDisease.replace("?", np.nan)
heartDisease = heartDisease.dropna()

# Rename last column as heartdisease
if "heartdisease" not in heartDisease.columns:
    heartDisease = heartDisease.rename(
        columns={heartDisease.columns[-1]: "heartdisease"}
    )

print("Dataset columns:")
print(heartDisease.columns.tolist())

# Create Bayesian Network
model = DiscreteBayesianNetwork([
    ("age", "trestbps"),
    ("age", "fbs"),
    ("sex", "trestbps"),
    ("exang", "trestbps"),
    ("trestbps", "heartdisease"),
    ("fbs", "heartdisease"),
    ("heartdisease", "restecg"),
    ("heartdisease", "thalach"),
    ("heartdisease", "chol")
])

# Learn CPDs
estimator = MaximumLikelihoodEstimator(model, heartDisease)
cpds = estimator.get_parameters()

for cpd in cpds:
    model.add_cpds(cpd)

print("\nBayesian Network created successfully.")
print("Model check:", model.check_model())

print("\nLearned CPDs:")
for cpd in model.get_cpds():
    print(cpd)
```

```python
from pgmpy.inference import VariableElimination

# Create inference object
heartDisease_infer = VariableElimination(model)

# Diagnosis using age = 63
print("\n--- Diagnosis using Age = 63 ---")

q1 = heartDisease_infer.query(
    variables=["heartdisease"],
    evidence={"age": 63}
)

print(q1)


# Diagnosis using cholesterol = 233
print("\n--- Diagnosis using Cholesterol = 233 ---")

q2 = heartDisease_infer.query(
    variables=["heartdisease"],
    evidence={"chol": 233}
)

print(q2)
print("Aditya Rade   BEAD23167")
```
